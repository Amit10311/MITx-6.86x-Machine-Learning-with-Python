````python
def norm(A, B):
    """
    Takes two Numpy column arrays, A and B, and returns the L2 norm of their
    sum.

    Arg:
      A - a Numpy array
      B - a Numpy array
    Returns:
      s - the L2 norm of A+B.
    """
    #Your code here
    import numpy as np
    
    # A = np.random.random([2,1])
    # B = np.random.random([2,1])
    s = np.linalg.norm(A+B)
    return s
    raise NotImplementedError
````

We review the simplest neural network here:



Screenshot from 2023-02-10 14-16-08.png

The output of the neural network, $z_1$ , is dependent on the inputs $x_1$ and $x_2$.

*  The importance of each of the inputs is given by values called $weights$.

*  There is one weight from each input to each output. We show this here:

Screenshot from 2023-02-10 14-17-27.png

The inputs are given by $x$, and the outputs are given by $z_1$. Here, $w_{11}$ is the weight of input 1 on output 1 (our only output in this case), and $w_{21}$ is the weight of input 2 on output 1. 

* In general, $w_{ij}$ represents the weight of input $i$ on output $j$.

The output, $z_1$, is given by $z_1 = f(w_{11}x_1 + w_{21}x_2)$

Screenshot from 2023-02-10 14-21-24.png

where $f$ is a specified nonlinear function, and it is usually the hyperbolic tangent function , $tanh$ .

If we express our inputs and weights as matrices, as shown here,



Screenshot from 2023-02-10 14-22-52.png

### **Neural Network**


````python
Output size: (1, 1)
````

````python
def neural_network(inputs, weights):
    """
     Takes an input vector and runs it through a 1-layer neural network
     with a given weight matrix and returns the output.

     Arg:
       inputs - 2 x 1 NumPy array
       weights - 2 x 1 NumPy array
     Returns (in this order):
       out - a 1 x 1 NumPy array, representing the output of the neural network
    """
    #Your code here
    from numpy.linalg.linalg import transpose
    W = np.transpose(weights)  
    Z1 = np.matmul(W,inputs)
    out = np.tanh(Z1)
    return out 
    raise NotImplementedError
````

````python
import numpy as np

def scalar_function(x, y):
    """
    Returns the f(x,y) defined in the problem statement.
    """
    #Your code here
    if x <= y:
        return x*y
    else:
        return x/y
        
    raise NotImplementedError
````

### **Vectorize function**


````python
import numpy as np

def vector_function(x, y):
    """
    Make sure vector_function can deal with vector input x,y 
    """
    #Your code here
    return np.vectorize(scalar_function)(x, y)   
    
````

````python
def get_sum_metrics(predictions, metrics=[]):
    for i in range(3):
        metrics.append(lambda x: x + i)

    sum_metrics = 0
    for metric in metrics:
        sum_metrics += metric(predictions)
   
    return sum_metrics


def main():
    print(get_sum_metrics(0))  # Should be (0 + 0) + (0 + 1) + (0 + 2) = 3
    print(get_sum_metrics(1))  # Should be (1 + 0) + (1 + 1) + (1 + 2) = 6
    print(get_sum_metrics(2))  # Should be (2 + 0) + (2 + 1) + (2 + 2) = 9
    print(get_sum_metrics(3, [lambda x: x]))  # Should be (3) + (3 + 0) + (3 + 1) + (3 + 2) = 15
    print(get_sum_metrics(0))  # Should be (0 + 0) + (0 + 1) + (0 + 2) = 3
    print(get_sum_metrics(1))  # Should be (1 + 0) + (1 + 1) + (1 + 2) = 6
    print(get_sum_metrics(2))  # Should be (2 + 0) + (2 + 1) + (2 + 2) = 9

if __name__ == "__main__":
    main()
````

One thing to note is that, because i is referenced in the lambda functions, its value will be remembered when the functions are called outside the for loop.

*  This means that each lambda function will return x + 2 instead of the expected x + 0, x + 1, and x + 2. 

* **To fix this issue**, you could wrap the lambda function creation in another function, or use a list comprehension instead of a loop:

````python
# def get_sum_metrics(predictions, metrics=[]):
#     metrics = [lambda x, i=i: x + i for i in range(3)]

def get_sum_metrics(predictions, metrics=[]):
    metrics = metrics + [ lambda x, i=i : x + i for i in range(3) ]

    sum_metrics = 0
    for metric in metrics:
        sum_metrics += metric(predictions)
    return sum_metrics

def main():
  print(get_sum_metrics(0))  # Should be (0 + 0) + (0 + 1) + (0 + 2) = 3
  print(get_sum_metrics(1))  # Should be (1 + 0) + (1 + 1) + (1 + 2) = 6
  print(get_sum_metrics(2))  # Should be (2 + 0) + (2 + 1) + (2 + 2) = 9
  print(get_sum_metrics(2, [lambda x: x]))  # Should be (3) + (3 + 0) + (3 + 1) + (3 + 2) = 15
  print(get_sum_metrics(0))  # Should be (0 + 0) + (0 + 1) + (0 + 2) = 3
  print(get_sum_metrics(1))  # Should be (1 + 0) + (1 + 1) + (1 + 2) = 6
  print(get_sum_metrics(2))  # Should be (2 + 0) + (2 + 1) + (2 + 2) = 9

if __name__ == "__main__":
    main()
````
